```python
import matplotlib.pyplot as plt 
x = [1, 2, 3, 4]
y = [10, 20, 30, 40]
plt.plot(x, y)  
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Line Graph')
plt.show()
```

```python
import matplotlib.pyplot as plt
categotories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
plt.bar(categotories, values)
plt.xlabel('Categories')        
plt.ylabel('Values')
plt.title('Bar Graph')
plt.show()
```

```python
import matplotlib.pyplot as plt
data= [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
plt.hist(data, bins=4)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y=[5, 7, 6, 8, 7]
plt.scatter(x, y)
plt.xlabel('x-axis')        
plt.ylabel('y-axis')
plt.title('Scatter Plot')
plt.show()
```

```python
import matplotlib.pyplot as plt
lables = ['A', 'B', 'C', 'D', 'E']
sizes = [20, 35, 25, 45, 15]
plt.pie(sizes, labels=lables, autopct='%1.1f%%')
plt.title('Pie Chart')  
plt.show()
```

```python
import matplotlib.pyplot as plt
x=['A', 'B', 'C']
y1=[3,5,2]
y2=[4,6,3]
plt.bar(x, y1, label='Data1')
plt.bar(x, y2, bottom=y1, label='Data2')
plt.title('Stacked Bar Graph')
plt.legend
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
plt.step(x, y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')    
plt.title('Step plot')
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
plt.fill_between(x, y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')        
plt.title('Area plot')
plt.show()
```

```python
import matplotlib.pyplot as plt
data = [10,20,30,40,50,100]
plt.boxplot(data)
plt.title('Box Plot')       
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [5,15,10,20]
plt.stem(x, y)
plt.xlabel('x-axis')        
plt.ylabel('y-axis')
plt.title('Stem Plot')
plt.show()
```
